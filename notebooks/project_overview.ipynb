# Unpaired single-cell ↔ bulk translation: what adversarial training buys without pairs

COMP 690-204 (Deep Generative Models) final project.

Single-cell RNA-seq of peripheral blood mononuclear cells (PBMC), aggregated per sample into a pseudobulk profile, and bulk RNA-seq of whole blood measure overlapping biology on different scales, and the two are separated by a large technical and biological gap. We study whether a cycle-consistent adversarial translator can learn to map between them **without paired samples**, and how much paired data, or composition supervision derived from it, adds. The AIFI Sound Life cohort provides 836 blood draws from 96 donors with both assays from the same draw. Models are trained on donor-disjoint halves of the training donors (single-cell profiles from one half, bulk from the other), and the held-out pairs serve only as the answer key for evaluation.

Shared code is in the `dgm/` package, with known-answer tests in `tests/`.

## Reproducing the results

1. Create the environment: `conda env create -f environment.yml`.
2. Rebuild the single-cell input files from the AIFI portal files with `python scripts/make_slims.py <portal directory>`, and place the whole-blood bulk file next to them in `data/raw/soundlife/` (see `README.md`).
3. Run the phase notebooks in order, top to bottom. Each prints the commit it ran on, and training is deterministic on CPU.

The submitted proposal is in `proposal/` (`proposal_final.tex`, Figure 1), with its errors and the design changes made since in `proposal/ERRATA.md`.
